# Q.ANT's NPU runs a pretrained UNet (Image segmentation)

```python
import matplotlib.pyplot as plt
import torch
import utils
from qant_unet import QAntUNet
```

### Prepare input and neural network data

1. Load pretrained UNet for brain MRI data of PyTorch repository 

```python
# load pretrained weights
w = torch.hub.load(
    "mateuszbuda/brain-segmentation-pytorch",
    "unet",
    in_channels=3,
    out_channels=1,
    init_features=32,
    pretrained=True,
).state_dict()

# Use weights to create a Q.ANT Native Computing SDK based implementation
qant_model = QAntUNet(w)
```


2. Load a image from the web

```python
url = "https://github.com/mateuszbuda/brain-segmentation-pytorch/raw/master/assets/TCGA_CS_4944.png"

image, input_batch = utils.get_image(url)
_ = plt.imshow(image)
```

## Neural network inference

```python
# The NN inference
output = qant_model(input_batch)

# simplify output type for testing
output_list = output.tolist()

# Converting output features to a mask
output_image = utils.outline(image.copy(), output[0, 0].numpy(), (255, 0, 0))

# Display mask
plt.title("U-Net for brain MRI (NPU)")
plt.imshow(output_image)
plt.show()
```
